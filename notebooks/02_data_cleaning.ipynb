# Phase 1.2 - Nettoyage automatique des donnÃ©es RetailSenseAI

Ce notebook nettoie un dataset e-commerce avant tout entraÃ®nement Machine Learning.

Il montre clairement :

1. l'Ã©tat **avant** nettoyage ;
2. le traitement des valeurs manquantes ;
3. la suppression des doublons ;
4. la dÃ©tection et le plafonnement raisonnÃ© des valeurs aberrantes ;
5. le retrait des colonnes inutiles pour le modÃ¨le ;
6. l'Ã©tat **aprÃ¨s** nettoyage et la sauvegarde du rÃ©sultat.

Le notebook est autonome : si le CSV RetailSenseAI n'existe pas, il le recrÃ©e avec une graine fixe.

```python
# Imports, chemins et configuration reproductible
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pandas.api.types import is_bool_dtype, is_numeric_dtype

RANDOM_SEED = 42
CUSTOMER_COUNT = 1_200
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "notebooks" / "data"
SOURCE_PATH = DATA_DIR / "retailsense_ecommerce_customers.csv"
CLEAN_PATH = DATA_DIR / "retailsense_ecommerce_customers_clean.csv"
DATA_DIR.mkdir(parents=True, exist_ok=True)

print("Source :", SOURCE_PATH)
print("RÃ©sultat nettoyÃ© :", CLEAN_PATH)
```

## 1. Chargement autonome du dataset

La fonction de secours reproduit le dataset e-commerce de la Phase 1 avec les mÃªmes rÃ¨gles mÃ©tier et les mÃªmes dÃ©fauts contrÃ´lÃ©s.

Elle n'est appelÃ©e que si le fichier source est absent. Le notebook peut donc Ãªtre lancÃ© seul dans un environnement neuf.

```python
def create_retailsense_dataset(path: Path, row_count: int = CUSTOMER_COUNT) -> None:
    """CrÃ©e un dataset e-commerce rÃ©aliste avec anomalies contrÃ´lÃ©es."""
    rng = np.random.default_rng(RANDOM_SEED)
    customer_id = [f"CUST-{index:06d}" for index in range(1, row_count + 1)]
    age = np.clip(rng.normal(39, 12, row_count).round(), 18, 80).astype(float)
    gender = rng.choice(["Femme", "Homme", "Non renseignÃ©"], row_count, p=[0.49, 0.48, 0.03])
    country = rng.choice(["France", "Belgique", "Suisse"], row_count, p=[0.68, 0.20, 0.12])
    cities = {
        "France": ["Paris", "Lyon", "Lille", "Bordeaux"],
        "Belgique": ["Bruxelles", "LiÃ¨ge", "Namur"],
        "Suisse": ["GenÃ¨ve", "Lausanne", "Zurich"],
    }
    city = [rng.choice(cities[value]) for value in country]
    loyalty = rng.choice(["Bronze", "Silver", "Gold", "Platinum"], row_count, p=[0.44, 0.31, 0.19, 0.06])
    loyalty_bonus = pd.Series(loyalty).map({"Bronze": 0, "Silver": 2, "Gold": 5, "Platinum": 9}).to_numpy()
    frequency = np.maximum(1, rng.poisson(4 + loyalty_bonus * 0.45, row_count)).astype(float)
    order_value = np.clip(rng.lognormal(4.25, 0.48, row_count), 15, 450)
    orders = np.maximum(1, (frequency * rng.uniform(1.5, 4.5, row_count)).round()).astype(int)
    spent = order_value * orders * rng.uniform(0.92, 1.08, row_count)
    discount = np.clip(rng.beta(2.2, 4.5, row_count), 0, 1)
    inactivity = np.clip(rng.exponential(38, row_count).round(), 0, 365).astype(int)
    returns = rng.binomial(orders, np.clip(0.03 + discount * 0.13, 0, 0.35))
    newsletter = rng.random(row_count) < 0.62
    signup = pd.Timestamp("2025-12-31") - pd.to_timedelta(rng.integers(30, 1_800, row_count), unit="D")
    churn_logit = -2 + 0.025 * inactivity - 0.13 * frequency - 0.00022 * spent + 0.9 * discount + 0.08 * returns - 0.42 * newsletter - 0.10 * loyalty_bonus
    churn = rng.random(row_count) < (1 / (1 + np.exp(-churn_logit)))

    frame = pd.DataFrame({
        "customer_id": customer_id,
        "age": age,
        "gender": gender,
        "city": city,
        "country": country,
        "acquisition_channel": rng.choice(["Organic", "Paid Search", "Social Media", "Referral", "Marketplace"], row_count),
        "loyalty_level": loyalty,
        "purchase_frequency": frequency,
        "average_order_value": order_value.round(2),
        "total_orders": orders,
        "total_spent": spent.round(2),
        "discount_usage": discount.round(3),
        "last_purchase_days": inactivity,
        "number_returns": returns,
        "payment_method": rng.choice(["Carte", "PayPal", "Virement", "Apple Pay"], row_count),
        "device": rng.choice(["Mobile", "Desktop", "Tablet"], row_count, p=[0.59, 0.34, 0.07]),
        "newsletter_subscribed": newsletter,
        "signup_date": signup,
        "record_source": "retailsense_demo",
        "churn": churn,
    })

    # Anomalies contrÃ´lÃ©es pour dÃ©montrer le nettoyage.
    for column, fraction in {"age": 0.025, "city": 0.020, "average_order_value": 0.018, "payment_method": 0.015}.items():
        indexes = rng.choice(frame.index, int(row_count * fraction), replace=False)
        frame.loc[indexes, column] = np.nan
    frame.loc[rng.choice(frame.index, 6, replace=False), "total_spent"] *= 8
    frame = pd.concat([frame, frame.sample(8, random_state=RANDOM_SEED)], ignore_index=True)
    frame.to_csv(path, index=False, date_format="%Y-%m-%d")


if not SOURCE_PATH.exists():
    create_retailsense_dataset(SOURCE_PATH)

data_before = pd.read_csv(SOURCE_PATH, parse_dates=["signup_date"])
print(f"{len(data_before):,} lignes x {len(data_before.columns)} colonnes chargÃ©es")
display(data_before.head(3))
```

## 2. Ã‰tat initial : AVANT nettoyage

Le rapport suivant mesure les problÃ¨mes sans modifier le DataFrame :

- cellules manquantes par colonne ;
- lignes entiÃ¨rement dupliquÃ©es ;
- colonnes constantes, donc sans information prÃ©dictive ;
- valeurs extrÃªmes selon la rÃ¨gle IQR.

Une valeur situÃ©e hors de $[Q1 - 1.5 \times IQR,\ Q3 + 1.5 \times IQR]$ est signalÃ©e comme aberrante. Elle n'est pas automatiquement une erreur : la dÃ©cision dÃ©pend du sens mÃ©tier.

```python
def iqr_bounds(series: pd.Series) -> tuple[float, float]:
    """Retourne les bornes basse et haute de la rÃ¨gle IQR."""
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    return float(q1 - 1.5 * iqr), float(q3 + 1.5 * iqr)


def count_outliers(series: pd.Series) -> int:
    lower, upper = iqr_bounds(series.dropna())
    return int(((series < lower) | (series > upper)).sum())


missing_before = data_before.isna().sum().sort_values(ascending=False)
duplicates_before = int(data_before.duplicated().sum())
constant_columns = [column for column in data_before if data_before[column].nunique(dropna=False) <= 1]
monetary_columns = ["average_order_value", "total_spent"]
outliers_before = {column: count_outliers(data_before[column]) for column in monetary_columns}

initial_report = pd.DataFrame({
    "indicateur": ["lignes", "colonnes", "cellules manquantes", "doublons", "colonnes constantes"],
    "valeur": [len(data_before), len(data_before.columns), int(missing_before.sum()), duplicates_before, len(constant_columns)],
})

display(initial_report)
print("Colonnes constantes :", constant_columns)
print("Valeurs aberrantes monÃ©taires :", outliers_before)
display(missing_before[missing_before > 0].rename("valeurs_manquantes").to_frame())
```

```python
# Visualisation des problÃ¨mes avant nettoyage
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

missing_before[missing_before > 0].sort_values().plot.barh(
    ax=axes[0], color="#d95f59", title="Valeurs manquantes - avant"
)
axes[0].set_xlabel("Nombre de cellules")
axes[0].set_ylabel("Colonne")

data_before[monetary_columns].plot.box(
    ax=axes[1], showfliers=True, title="Variables monÃ©taires - avant"
)
axes[1].set_ylabel("Montant")

plt.tight_layout()
plt.show()
```

## 3. StratÃ©gie de nettoyage automatique

Les rÃ¨gles restent simples et explicables :

| ProblÃ¨me | Traitement |
|---|---|
| Doublon complet | conserver la premiÃ¨re ligne |
| NumÃ©rique manquant | remplacer par la mÃ©diane |
| CatÃ©gorie manquante | remplacer par la modalitÃ© la plus frÃ©quente |
| Colonne constante | supprimer |
| Montant hors bornes IQR | plafonner Ã  la borne la plus proche |
| Identifiant client | conserver pour la traÃ§abilitÃ©, exclure du futur modÃ¨le |

Le plafonnement, aussi appelÃ© *winsorisation*, Ã©vite qu'une poignÃ©e d'erreurs de saisie domine les statistiques sans supprimer les clients concernÃ©s.

```python
data_clean = data_before.copy(deep=True)
cleaning_log = []

# 1. Doublons
rows_before = len(data_clean)
data_clean = data_clean.drop_duplicates().reset_index(drop=True)
cleaning_log.append({"Ã©tape": "doublons supprimÃ©s", "nombre": rows_before - len(data_clean)})

# 2. Colonnes sans information
removed_columns = [column for column in data_clean if data_clean[column].nunique(dropna=False) <= 1]
data_clean = data_clean.drop(columns=removed_columns)
cleaning_log.append({"Ã©tape": "colonnes constantes supprimÃ©es", "nombre": len(removed_columns)})

# 3. Imputation selon le type
numeric_features = [
    column for column in data_clean.select_dtypes(include="number").columns
    if column != "churn"
]
categorical_features = [
    column for column in data_clean.select_dtypes(include=["object", "string"]).columns
    if column != "customer_id"
]

for column in numeric_features:
    missing_count = int(data_clean[column].isna().sum())
    if missing_count:
        data_clean[column] = data_clean[column].fillna(data_clean[column].median())
        cleaning_log.append({"Ã©tape": f"{column} : mÃ©diane", "nombre": missing_count})

for column in categorical_features:
    missing_count = int(data_clean[column].isna().sum())
    if missing_count:
        mode = data_clean[column].mode(dropna=True).iloc[0]
        data_clean[column] = data_clean[column].fillna(mode)
        cleaning_log.append({"Ã©tape": f"{column} : modalitÃ© frÃ©quente", "nombre": missing_count})

# 4. Plafonnement des variables monÃ©taires
outlier_limits = {}
for column in monetary_columns:
    lower, upper = iqr_bounds(data_clean[column])
    changed_count = count_outliers(data_clean[column])
    data_clean[column] = data_clean[column].clip(lower=lower, upper=upper)
    outlier_limits[column] = {"minimum": round(lower, 2), "maximum": round(upper, 2)}
    cleaning_log.append({"Ã©tape": f"{column} : valeurs plafonnÃ©es", "nombre": changed_count})

cleaning_report = pd.DataFrame(cleaning_log)
display(cleaning_report)
print("Bornes IQR appliquÃ©es :", outlier_limits)
print("Colonne identifiant conservÃ©e :", "customer_id" in data_clean.columns)
```

## 4. Ã‰tat final : APRÃˆS nettoyage

La validation finale ne se contente pas d'afficher le rÃ©sultat : elle utilise des assertions exÃ©cutables.

Si une rÃ¨gle essentielle n'est plus respectÃ©e aprÃ¨s une modification future, le notebook s'arrÃªtera avec une erreur explicite au lieu de produire silencieusement un dataset invalide.

```python
missing_after = int(data_clean.isna().sum().sum())
duplicates_after = int(data_clean.duplicated().sum())
outliers_after = {column: count_outliers(data_clean[column]) for column in monetary_columns}

validation_results = {
    "aucune valeur manquante": missing_after == 0,
    "aucun doublon": duplicates_after == 0,
    "identifiants clients uniques": data_clean["customer_id"].is_unique,
    "cible churn conservÃ©e": "churn" in data_clean.columns,
    "colonne constante retirÃ©e": "record_source" not in data_clean.columns,
    "dates correctement typÃ©es": pd.api.types.is_datetime64_any_dtype(data_clean["signup_date"]),
    "valeurs monÃ©taires non nÃ©gatives": bool((data_clean[monetary_columns] >= 0).all().all()),
    "valeurs aberrantes monÃ©taires traitÃ©es": sum(outliers_after.values()) == 0,
}

validation_table = pd.DataFrame(
    [{"rÃ¨gle": rule, "validÃ©e": valid} for rule, valid in validation_results.items()]
)
display(validation_table)

assert all(validation_results.values()), "Le dataset nettoyÃ© ne respecte pas toutes les rÃ¨gles"
print("Toutes les validations sont rÃ©ussies.")
```
