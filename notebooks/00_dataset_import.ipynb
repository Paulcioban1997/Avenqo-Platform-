# 00 - Importation du dataset RetailSenseAI

## Objectif

Ce notebook crÃ©e puis importe un dataset rÃ©aliste reprÃ©sentant les clients d'une entreprise e-commerce. Il constitue le point de dÃ©part R&D du AI Engine Avenqo.

Le dataset est **synthÃ©tique, dÃ©terministe et non mÃ©dical** : les mÃªmes paramÃ¨tres et la mÃªme graine produisent toujours les mÃªmes donnÃ©es. Il contient volontairement quelques valeurs manquantes, doublons et valeurs extrÃªmes qui seront traitÃ©s dans le notebook `02_data_cleaning.ipynb`.

```python
# BibliothÃ¨ques et configuration reproductible
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

RANDOM_SEED = 42
CUSTOMER_COUNT = 1_200
rng = np.random.default_rng(RANDOM_SEED)

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_DIR = PROJECT_ROOT / 'notebooks' / 'data'
DATA_PATH = DATA_DIR / 'retailsense_ecommerce_customers.csv'
DATA_DIR.mkdir(parents=True, exist_ok=True)

print(f'Graine alÃ©atoire : {RANDOM_SEED}')
print(f'Fichier du dataset : {DATA_PATH}')
```

## 1. Contexte mÃ©tier

RetailSenseAI souhaite comprendre le comportement d'achat et anticiper le dÃ©part des clients. Une ligne reprÃ©sente un client.

Les variables couvrent quatre familles :

- **profil** : Ã¢ge, genre, ville, pays ;
- **acquisition et fidÃ©litÃ©** : canal, niveau de fidÃ©litÃ©, newsletter ;
- **comportement d'achat** : frÃ©quence, commandes, montant moyen, dÃ©penses ;
- **relation rÃ©cente** : jours depuis le dernier achat, retours, remises ;
- **cible** : `churn`, Ã©gale Ã  `True` lorsqu'un client est considÃ©rÃ© perdu.

```python
# GÃ©nÃ©ration d'un dataset e-commerce cohÃ©rent
customer_id = np.array([f'CUST-{index:06d}' for index in range(1, CUSTOMER_COUNT + 1)])
age = np.clip(rng.normal(39, 12, CUSTOMER_COUNT).round(), 18, 80).astype(float)
gender = rng.choice(['Femme', 'Homme', 'Non renseignÃ©'], CUSTOMER_COUNT, p=[0.49, 0.48, 0.03])
country = rng.choice(['France', 'Belgique', 'Suisse'], CUSTOMER_COUNT, p=[0.68, 0.20, 0.12])
cities = {
    'France': ['Paris', 'Lyon', 'Lille', 'Bordeaux'],
    'Belgique': ['Bruxelles', 'LiÃ¨ge', 'Namur'],
    'Suisse': ['GenÃ¨ve', 'Lausanne', 'Zurich'],
}
city = np.array([rng.choice(cities[value]) for value in country], dtype=object)
acquisition_channel = rng.choice(
    ['Organic', 'Paid Search', 'Social Media', 'Referral', 'Marketplace'],
    CUSTOMER_COUNT, p=[0.28, 0.24, 0.20, 0.16, 0.12],
)
loyalty_level = rng.choice(['Bronze', 'Silver', 'Gold', 'Platinum'], CUSTOMER_COUNT, p=[0.44, 0.31, 0.19, 0.06])
loyalty_bonus = pd.Series(loyalty_level).map({'Bronze': 0, 'Silver': 2, 'Gold': 5, 'Platinum': 9}).to_numpy()
purchase_frequency = np.maximum(1, rng.poisson(4 + loyalty_bonus * 0.45, CUSTOMER_COUNT)).astype(float)
average_order_value = np.clip(rng.lognormal(4.25, 0.48, CUSTOMER_COUNT), 15, 450)
total_orders = np.maximum(1, (purchase_frequency * rng.uniform(1.5, 4.5, CUSTOMER_COUNT)).round()).astype(int)
total_spent = average_order_value * total_orders * rng.uniform(0.92, 1.08, CUSTOMER_COUNT)
discount_usage = np.clip(rng.beta(2.2, 4.5, CUSTOMER_COUNT), 0, 1)
last_purchase_days = np.clip(rng.exponential(38, CUSTOMER_COUNT).round(), 0, 365).astype(int)
return_probability = np.clip(0.03 + discount_usage * 0.13, 0, 0.35)
number_returns = rng.binomial(total_orders, return_probability)
payment_method = rng.choice(['Carte', 'PayPal', 'Virement', 'Apple Pay'], CUSTOMER_COUNT, p=[0.52, 0.25, 0.13, 0.10])
device = rng.choice(['Mobile', 'Desktop', 'Tablet'], CUSTOMER_COUNT, p=[0.59, 0.34, 0.07])
newsletter_subscribed = rng.random(CUSTOMER_COUNT) < 0.62
signup_date = pd.Timestamp('2025-12-31') - pd.to_timedelta(rng.integers(30, 1_800, CUSTOMER_COUNT), unit='D')

# La probabilitÃ© de churn dÃ©pend de signaux mÃ©tier, puis reÃ§oit un bruit contrÃ´lÃ©.
churn_logit = (
    -2.0
    + 0.025 * last_purchase_days
    - 0.13 * purchase_frequency
    - 0.00022 * total_spent
    + 0.90 * discount_usage
    + 0.08 * number_returns
    - 0.42 * newsletter_subscribed.astype(int)
    - 0.10 * loyalty_bonus
)
churn_probability = 1 / (1 + np.exp(-churn_logit))
churn = rng.random(CUSTOMER_COUNT) < churn_probability

raw_data = pd.DataFrame({
    'customer_id': customer_id, 'age': age, 'gender': gender, 'city': city, 'country': country,
    'acquisition_channel': acquisition_channel, 'loyalty_level': loyalty_level,
    'purchase_frequency': purchase_frequency, 'average_order_value': average_order_value.round(2),
    'total_orders': total_orders, 'total_spent': total_spent.round(2),
    'discount_usage': discount_usage.round(3), 'last_purchase_days': last_purchase_days,
    'number_returns': number_returns, 'payment_method': payment_method, 'device': device,
    'newsletter_subscribed': newsletter_subscribed, 'signup_date': signup_date,
    'record_source': 'retailsense_demo', 'churn': churn,
})

# DÃ©fauts contrÃ´lÃ©s destinÃ©s aux notebooks de qualitÃ© et nettoyage.
for column, fraction in {'age': 0.025, 'city': 0.020, 'average_order_value': 0.018, 'payment_method': 0.015}.items():
    missing_index = rng.choice(raw_data.index, size=int(CUSTOMER_COUNT * fraction), replace=False)
    raw_data.loc[missing_index, column] = np.nan
raw_data.loc[rng.choice(raw_data.index, 6, replace=False), 'total_spent'] *= 8
raw_data = pd.concat([raw_data, raw_data.sample(8, random_state=RANDOM_SEED)], ignore_index=True)

raw_data.to_csv(DATA_PATH, index=False, date_format='%Y-%m-%d')
print(f'{len(raw_data):,} lignes Ã©crites dans le CSV')
```

## 2. Importation du fichier

Nous relisons maintenant le CSV depuis le disque. `signup_date` est convertie explicitement en datetime afin de conserver sa signification temporelle. Cette sÃ©paration gÃ©nÃ©ration â†’ sauvegarde â†’ import reproduit un upload utilisateur.

```python
data = pd.read_csv(
    DATA_PATH,
    parse_dates=['signup_date'],
    dtype={'customer_id': 'string'},
)

print(f'Dimensions : {data.shape[0]:,} lignes x {data.shape[1]} colonnes')
display(data.head())
```

## 3. Exploration et statistiques

Les statistiques numÃ©riques permettent de contrÃ´ler les ordres de grandeur. Le tableau de qualitÃ© donne les types, le nombre de valeurs manquantes et la cardinalitÃ© de chaque colonne.

```python
quality_summary = pd.DataFrame({
    'type': data.dtypes.astype(str),
    'non_nulles': data.notna().sum(),
    'manquantes': data.isna().sum(),
    'manquantes_pct': (data.isna().mean() * 100).round(2),
    'valeurs_uniques': data.nunique(dropna=True),
})

print('Statistiques numÃ©riques :')
display(data.describe().T.round(2))
print('QualitÃ© et types :')
display(quality_summary)
print('Doublons complets :', int(data.duplicated().sum()))
print('Taux de churn :', f"{data['churn'].mean():.1%}")
```

## 4. PremiÃ¨res visualisations mÃ©tier

Les graphiques rÃ©pondent Ã  quatre questions initiales :

1. la cible churn est-elle dÃ©sÃ©quilibrÃ©e ?
2. quels niveaux de fidÃ©litÃ© sont reprÃ©sentÃ©s ?
3. l'inactivitÃ© rÃ©cente diffÃ©rencie-t-elle les clients perdus ?
4. comment les dÃ©penses totales sont-elles distribuÃ©es ?

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

data['churn'].value_counts().sort_index().rename({False: 'Actif', True: 'Churn'}).plot(
    kind='bar', ax=axes[0, 0], color=['#168aad', '#d1495b'], title='RÃ©partition de la cible churn'
)
axes[0, 0].set_ylabel('Nombre de clients')
axes[0, 0].tick_params(axis='x', rotation=0)

loyalty_order = ['Bronze', 'Silver', 'Gold', 'Platinum']
data['loyalty_level'].value_counts().reindex(loyalty_order).plot(
    kind='bar', ax=axes[0, 1], color='#f4a261', title='Clients par niveau de fidÃ©litÃ©'
)
axes[0, 1].tick_params(axis='x', rotation=0)

data.boxplot(column='last_purchase_days', by='churn', ax=axes[1, 0], grid=False)
axes[1, 0].set_title('InactivitÃ© selon le churn')
axes[1, 0].set_xlabel('Churn')
axes[1, 0].set_ylabel('Jours depuis le dernier achat')

axes[1, 1].hist(data['total_spent'].dropna(), bins=35, color='#2a9d8f', edgecolor='white')
axes[1, 1].set_title('Distribution des dÃ©penses totales')
axes[1, 1].set_xlabel('DÃ©penses totales')
axes[1, 1].set_ylabel('Clients')

fig.suptitle('RetailSenseAI - Exploration initiale', fontsize=16)
plt.tight_layout()
plt.show()
```
