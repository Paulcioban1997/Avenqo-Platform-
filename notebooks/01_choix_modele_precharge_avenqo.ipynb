L'AI Engine doit seulement :

1. analyser le fichier chargÃ© ;

3. choisir `classification`, `regression` ou `clustering` ;
4. charger le pipeline prÃ©entraÃ®nÃ© correspondant ;

Ce notebook explique cette dÃ©cision. Il ne contient aucun MLP et n'entraÃ®ne aucun modÃ¨le.

````python
from dataclasses import dataclass
from enum import StrEnum
import pandas as pd
from pandas.api.types import is_numeric_dtype

class TypeTache(StrEnum):
    CLASSIFICATION = "classification"
    REGRESSION = "regression"
    CLUSTERING = "clustering"

@dataclass(frozen=True)
class Decision:
    type_tache: TypeTache
    raison: str


def choisir_tache(data: pd.DataFrame, cible: str | None = None) -> Decision:
    """ReprÃ©sentation pÃ©dagogique de la dÃ©cision future de l'AI Engine."""
    if cible is None:
        return Decision(TypeTache.CLUSTERING, "Aucune cible fournie")
    if cible not in data:
        raise ValueError(f"Colonne cible absente: {cible}")
    valeurs = data[cible].dropna()
    if not is_numeric_dtype(valeurs) or valeurs.nunique() <= 20:
        return Decision(TypeTache.CLASSIFICATION, "Cible catÃ©gorielle ou discrÃ¨te")
    return Decision(TypeTache.REGRESSION, "Cible numÃ©rique continue")
````

````python
classification_data = pd.DataFrame({"age": [25, 40, 52], "churn": [0, 1, 0]})
regression_data = pd.DataFrame({
    "quantity": range(30),
    "revenue": [index * 1.37 for index in range(30)],
})
clustering_data = pd.DataFrame({"recency": [2, 15, 80], "frequency": [20, 7, 1]})

examples = [
    ("PrÃ©dire le churn", choisir_tache(classification_data, "churn")),
    ("PrÃ©voir le revenu", choisir_tache(regression_data, "revenue")),
    ("Segmenter les clients", choisir_tache(clustering_data)),
]

for situation, decision in examples:
    print(f"{situation:24} -> {decision.type_tache.value:14} | {decision.raison}")
````
