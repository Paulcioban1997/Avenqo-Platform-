## 1. Importation du dataset

Pour que le notebook soit autonome, nous partons du dataset public `breast_cancer` inclus dans scikit-learn. Nous l'enrichissons avec :

- une colonne catÃ©gorielle `centre` ;
- quelques valeurs manquantes ;
- une ligne dupliquÃ©e.

Le dataset est ensuite Ã©crit en CSV puis **rÃ©importÃ© depuis le disque**. Cette Ã©tape reproduit le comportement d'un fichier chargÃ© dans Avenqo.

```python
# CrÃ©ation du fichier d'entrÃ©e, puis import comme un vrai upload
source = load_breast_cancer(as_frame=True)
raw_data = source.frame.copy()
raw_data = raw_data.rename(columns={"target": "diagnostic"})
raw_data["centre"] = np.where(raw_data.index % 3 == 0, "Bruxelles", "Paris")

# Valeurs manquantes injectÃ©es uniquement pour tester le preprocessing.
missing_rows = raw_data.sample(frac=0.04, random_state=RANDOM_SEED).index
raw_data.loc[missing_rows, "mean radius"] = np.nan
raw_data.loc[missing_rows[:10], "centre"] = np.nan

# Une ligne dupliquÃ©e permet de valider le nettoyage explicite.
raw_data = pd.concat([raw_data, raw_data.iloc[[0]]], ignore_index=True)
raw_data.to_csv(DATA_PATH, index=False)

data = pd.read_csv(DATA_PATH)
print(f"Fichier importÃ© : {DATA_PATH}")
print(f"Dimensions brutes : {data.shape[0]} lignes x {data.shape[1]} colonnes")
display(data.head())
```

## 2. Diagnostic et dÃ©tection automatique des colonnes

La cible est `diagnostic`. Elle ne doit jamais entrer dans la liste des features.

La rÃ¨gle simple utilisÃ©e ici :

- types numÃ©riques pandas â†’ colonnes numÃ©riques ;
- tous les autres types â†’ colonnes catÃ©gorielles.

Avant de construire le Pipeline, nous mesurons aussi les valeurs manquantes, doublons et la distribution de la cible.

```python
TARGET_COLUMN = "diagnostic"
feature_columns = [column for column in data.columns if column != TARGET_COLUMN]
numerical_columns = tuple(data[feature_columns].select_dtypes(include="number").columns)
categorical_columns = tuple(
    column for column in feature_columns if column not in numerical_columns
)

quality_report = pd.DataFrame(
    {
        "type": data.dtypes.astype(str),
        "valeurs_manquantes": data.isna().sum(),
        "taux_manquant_pct": (data.isna().mean() * 100).round(2),
        "valeurs_uniques": data.nunique(dropna=True),
    }
)

print("Colonnes numÃ©riques :", len(numerical_columns))
print("Colonnes catÃ©gorielles :", categorical_columns)
print("Doublons dÃ©tectÃ©s :", int(data.duplicated().sum()))
print("Distribution de la cible :")
display(data[TARGET_COLUMN].value_counts().rename_axis("classe").to_frame("nombre"))
display(quality_report.query("valeurs_manquantes > 0"))
```

## 3. Nettoyage et feature engineering

Nous rÃ©alisons uniquement les transformations qui peuvent Ãªtre calculÃ©es sans apprendre d'information globale :

- suppression des doublons ;
- suppression des lignes dont la cible est absente ;
- crÃ©ation de `compactness_ratio` Ã  partir de deux colonnes de la mÃªme ligne.

Les mÃ©dianes, modalitÃ©s frÃ©quentes, catÃ©gories One-Hot et paramÃ¨tres du scaler seront appris **aprÃ¨s la sÃ©paration train/test**, Ã  l'intÃ©rieur du Pipeline.

```python
clean_data = data.drop_duplicates().dropna(subset=[TARGET_COLUMN]).copy()
clean_data["compactness_ratio"] = (
    clean_data["mean compactness"]
    / clean_data["mean perimeter"].replace(0, np.nan)
)

X = clean_data.drop(columns=[TARGET_COLUMN])
y = clean_data[TARGET_COLUMN].astype(int)

numerical_columns = tuple(X.select_dtypes(include="number").columns)
categorical_columns = tuple(column for column in X if column not in numerical_columns)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_SEED,
    stratify=y,
)

print(f"AprÃ¨s nettoyage : {len(clean_data)} lignes")
print(f"Train : {X_train.shape} | Test : {X_test.shape}")
print("Nouvelle feature : compactness_ratio")
print("Cible retirÃ©e des features :", TARGET_COLUMN not in X.columns)
```

## 4. Preprocessing et Pipelines sklearn

Le `ColumnTransformer` applique deux branches :

### Variables numÃ©riques

1. imputation par la mÃ©diane ;
2. standardisation avec `StandardScaler`.

### Variables catÃ©gorielles

1. imputation par la modalitÃ© la plus frÃ©quente ;
2. encodage One-Hot avec tolÃ©rance des catÃ©gories inconnues.

Chaque modÃ¨le est placÃ© Ã  la fin d'un `Pipeline`. GridSearchCV peut ainsi rÃ©apprendre correctement le preprocessing dans chaque fold de validation croisÃ©e.

```python
numerical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numerical", numerical_pipeline, list(numerical_columns)),
        ("categorical", categorical_pipeline, list(categorical_columns)),
    ],
    remainder="drop",
)

candidates = {
    "logistic_regression": (
        LogisticRegression(max_iter=2000, random_state=RANDOM_SEED),
        {
            "model__C": [0.1, 1.0, 10.0],
            "model__class_weight": [None, "balanced"],
        },
    ),
    "random_forest": (
        RandomForestClassifier(random_state=RANDOM_SEED, n_jobs=1),
        {
            "model__n_estimators": [100, 200,300,400,500],
            "model__max_depth": [None, 8,16,25,30,50],
            "model__min_samples_split": [2, 5, 10],
            "model__min_samples_leaf": [1, 2, 4],
            "model__class_weight": [None, "balanced"],
            "model__max_features": ["sqrt", "log2"],
            "model__criterion": ["gini", "entropy"],
            "model__bootstrap": [True, False],
        },
    ),
    "adaboost": (
        AdaBoostClassifier(random_state=RANDOM_SEED, n_jobs=1),
        {
            "model__n_estimators": [50, 100],
            "model__learning_rate": [0.5, 1.0],
            "model__algorithm": ["SAMME", "SAMME.R"],
            "model__base_estimator": [None, RandomForestClassifier(random_state=RANDOM_SEED)]
        },
    ),
    "gradient_boosting": (
        GradientBoostingClassifier(random_state=RANDOM_SEED),
        {
            "model__n_estimators": [100, 200],
            "model__learning_rate": [0.01, 0.1, 0.2],
            "model__max_depth": [3, 5, 7],
            "model__min_samples_split": [2, 5, 10],
            "model__min_samples_leaf": [1, 2, 4],
        },
    ),

    "xgboost": (
        XGBClassifier(random_state=RANDOM_SEED, n_jobs=1, use_label_encoder=False, eval_metric="logloss"),
        {
            "model__n_estimators": [100, 200],
            "model__learning_rate": [0.01, 0.1, 0.2],
            "model__max_depth": [3, 5, 7],
            "model__subsample": [0.8, 1.0],
            "model__colsample_bytree": [0.8, 1.0],
            "model__gamma": [0, 0.1, 0.2],
        },
    ),

}

print("Ã‰tapes numÃ©riques :", numerical_pipeline.steps)
print("Ã‰tapes catÃ©gorielles :", categorical_pipeline.steps)
print("ModÃ¨les candidats :", list(candidates))
```

## 5. Recherche d'hyperparamÃ¨tres

Nous lanÃ§ons une `GridSearchCV` indÃ©pendante pour chaque famille de modÃ¨le.

- scoring principal : `roc_auc` ;
- validation croisÃ©e stratifiÃ©e gÃ©rÃ©e par scikit-learn ;
- `refit=True` : le meilleur pipeline est rÃ©entraÃ®nÃ© automatiquement sur tout le train ;
- le jeu de test reste totalement invisible jusqu'Ã  l'Ã©valuation finale.

Pour une grille trÃ¨s grande, cette cellule pourra Ãªtre remplacÃ©e par `RandomizedSearchCV` sans modifier le reste du notebook.

```python
searches = {}
comparison_rows = []
training_started = time.perf_counter()

for model_name, (estimator, parameter_grid) in candidates.items():
    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", estimator),
        ]
    )
    search = GridSearchCV(
        estimator=pipeline,
        param_grid=parameter_grid,
        scoring="roc_auc",
        cv=5,
        n_jobs=-1,
        refit=True,
        return_train_score=True,
        error_score="raise",
    )
    search.fit(X_train, y_train)
    searches[model_name] = search
    comparison_rows.append(
        {
            "modele": model_name,
            "meilleur_roc_auc_cv": search.best_score_,
            "best_params": search.best_params_,
        }
    )

training_duration_seconds = time.perf_counter() - training_started
comparison = pd.DataFrame(comparison_rows).sort_values(
    "meilleur_roc_auc_cv", ascending=False
)
best_model_name = comparison.iloc[0]["modele"]
best_search = searches[best_model_name]
best_estimator = best_search.best_estimator_

print(f"DurÃ©e totale : {training_duration_seconds:.2f} secondes")
display(comparison)
print("\nModÃ¨le sÃ©lectionnÃ© :", best_model_name)
print("best_params_ :", best_search.best_params_)
print("best_estimator_ :")
print(best_estimator)
```
