# 02 - Isolation des donnÃ©es par compagnie

Chaque compagnie connectÃ©e Ã  Avenqo possÃ¨de son propre espace de donnÃ©es.

Chaque fichier chargÃ© reÃ§oit aussi son propre identifiant et son propre emplacement :

```text
var/tenants/{company_id}/datasets/{dataset_id}/dataset.db
```

Ainsi, deux compagnies peuvent charger un fichier portant le mÃªme nom sans partager leurs donnÃ©es. Ce notebook dÃ©montre cette rÃ¨gle avec deux bases SQLite temporaires.

````python
from pathlib import Path
from tempfile import TemporaryDirectory
from uuid import UUID, uuid4
import sqlite3
import pandas as pd

COMPAGNIE_A = UUID("00000000-0000-0000-0000-000000000001")
COMPAGNIE_B = UUID("00000000-0000-0000-0000-000000000002")


def chemin_base(racine: Path, compagnie_id: UUID, dataset_id: UUID) -> Path:
    """Chaque chemin contient obligatoirement la compagnie et le fichier."""
    return racine / str(compagnie_id) / "datasets" / str(dataset_id) / "dataset.db"


def enregistrer_fichier(racine: Path, compagnie_id: UUID, data: pd.DataFrame) -> tuple[UUID, Path]:
    dataset_id = uuid4()
    destination = chemin_base(racine, compagnie_id, dataset_id)
    destination.parent.mkdir(parents=True, exist_ok=True)
    connexion = sqlite3.connect(destination)
    try:
        data.to_sql("data", connexion, index=False, if_exists="fail")
        connexion.commit()
    finally:
        connexion.close()
    return dataset_id, destination
````

````python
with TemporaryDirectory() as dossier_temporaire:
    racine = Path(dossier_temporaire)
    ventes_a = pd.DataFrame({"produit": ["A", "B"], "montant": [100, 250]})
    ventes_b = pd.DataFrame({"produit": ["SECRET"], "montant": [9999]})

    dataset_a, base_a = enregistrer_fichier(racine, COMPAGNIE_A, ventes_a)
    dataset_b, base_b = enregistrer_fichier(racine, COMPAGNIE_B, ventes_b)

    print("Base compagnie A:", base_a.relative_to(racine))
    print("Base compagnie B:", base_b.relative_to(racine))
    print("Bases diffÃ©rentes:", base_a != base_b)

    connexion_a = sqlite3.connect(base_a)
    connexion_b = sqlite3.connect(base_b)
    try:
        donnees_a = pd.read_sql_query("SELECT * FROM data", connexion_a)
        donnees_b = pd.read_sql_query("SELECT * FROM data", connexion_b)
    finally:
        connexion_a.close()
        connexion_b.close()

    display(donnees_a)
    display(donnees_b)
````

## RÃ¨gles obligatoires en production

- `company_id` vient de l'utilisateur authentifiÃ©, jamais d'un champ libre du formulaire.
- Toute lecture reconstruit le chemin avec ce `company_id` vÃ©rifiÃ©.
- Un identifiant de dataset appartenant Ã  une autre compagnie retourne Â« introuvable Â».
- Le fichier original, la base dÃ©rivÃ©e, les modÃ¨les et les Experiments sont isolÃ©s.
- Les sauvegardes et la suppression d'une compagnie suivent la mÃªme arborescence.

La base SQL centrale conserve seulement le catalogue (`datasets.company_id`, nom, statut, dimensions). Les lignes privÃ©es du fichier restent dans l'espace isolÃ© de la compagnie.

# Avenqo - Pipeline Deep Learning TensorFlow/Keras

Ce notebook entraÃ®ne un **vrai MLP de rÃ©gression** avec le moteur partagÃ© Avenqo. Il montre le preprocessing mixte, les callbacks `EarlyStopping` et `ReduceLROnPlateau`, les courbes d'apprentissage, les mÃ©triques, la sauvegarde `.keras` et la crÃ©ation automatique d'un `ExperimentRun`.

Le dataset est synthÃ©tique et reproductible : vous pouvez lancer **Run All**.

````python
# Cette cellule permet d'importer Avenqo depuis le dossier notebooks/ ou la racine.
from pathlib import Path
import sys
from uuid import UUID

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

from shared.ai_engine.contracts import DatasetArtifact, DetectedSchema, TenantContext
from shared.ai_engine.experiments import (
    DataPreparationRecord,
    DatasetSnapshot,
    InMemoryExperimentRepository,
    ReproducibilityRecord,
    SearchMethod,
)
from shared.ai_engine.model_registry.repository import FileSystemModelRepository
from shared.ai_engine.training import TrainingRunContext, TrainingService
from shared.ai_engine.training.pipeline_builder import build_preprocessor, detect_feature_columns

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)
MODEL_ROOT = PROJECT_ROOT / "var" / "notebook_models" / "deep_learning"
print(f"TensorFlow : {tf.__version__}")
print(f"ModÃ¨les du notebook : {MODEL_ROOT}")
````

## 1. Contexte Avenqo et ExperimentRun

Le rÃ©seau neuronal reste partagÃ© par tous les modules. Le module fournit le dataset et les mÃ©tadonnÃ©es reproductibles ; `TrainingService` entraÃ®ne, Ã©value, sauvegarde et journalise automatiquement le Run.

````python
TENANT = TenantContext(UUID("00000000-0000-0000-0000-000000000001"))
DATASET = DatasetArtifact(
    tenant=TENANT,
    module_code="retail",
    task_code="demand_neural_network",
    uri="memory://demand-neural-network",
    schema=DetectedSchema(tables={}),
)
experiments = InMemoryExperimentRepository()
training_service = TrainingService(
    FileSystemModelRepository(MODEL_ROOT),
    experiments,
)


def build_run_context(data: pd.DataFrame) -> TrainingRunContext:
    feature_data = data.drop(columns=["demand"])
    columns = detect_feature_columns(feature_data)
    return TrainingRunContext(
        dataset=DatasetSnapshot(
            dataset_id=UUID("20000000-0000-0000-0000-000000000001"),
            version="notebook-v1",
            fingerprint="demo:demand-neural-network:seed-42",
            uri="memory://demand-neural-network",
            row_count=len(data),
            column_count=len(data.columns),
            numerical_columns=columns.numerical,
            categorical_columns=columns.categorical,
        ),
        preparation=DataPreparationRecord(
            created_columns=("demand",),
            feature_engineering=("demand synthÃ©tique avec saisonnalitÃ©",),
            imputation_strategy="median/mode",
            encoding_strategy="one_hot",
            scaling_strategy="standard_scaler",
        ),
        reproducibility=ReproducibilityRecord(
            random_seed=RANDOM_SEED,
            split_strategy="train_test_split",
            split_parameters={"test_size": 0.2},
            python_version=sys.version.split()[0],
            library_versions={"tensorflow": tf.__version__, "pandas": pd.__version__},
            code_version="notebook-demo-v1",
        ),
        search_method=SearchMethod.FIXED,
        parameter_spaces={},
    )
````

## 2. Dataset de demande

Le dataset combine deux nombres (`price`, `promotion`) et deux catÃ©gories (`channel`, `season`). La cible `demand` contient une relation non linÃ©aire et un bruit lÃ©ger.

````python
sample_count = 320
rng = np.random.default_rng(RANDOM_SEED)
price = rng.uniform(5, 80, sample_count)
promotion = rng.uniform(0, 0.4, sample_count)
channel = rng.choice(["web", "store"], sample_count)
season = rng.choice(["winter", "spring", "summer", "autumn"], sample_count)
season_effect = pd.Series(season).map({"winter": 8, "spring": 3, "summer": 12, "autumn": 5}).to_numpy()
channel_effect = np.where(channel == "web", 6, 0)
noise = rng.normal(0, 2, sample_count)

demand_data = pd.DataFrame(
    {
        "price": price,
        "promotion": promotion,
        "channel": channel,
        "season": season,
        "demand": 95 - 0.8 * price + 55 * promotion + season_effect + channel_effect + noise,
    }
)
demand_data.head()
````

````python
deep_features = demand_data.drop(columns=["demand"])
deep_columns = detect_feature_columns(deep_features)
deep_preprocessor = build_preprocessor(deep_columns)
transformed_preview = deep_preprocessor.fit_transform(deep_features.head(20))

print("Colonnes numÃ©riques   :", deep_columns.numerical)
print("Colonnes catÃ©gorielles:", deep_columns.categorical)
print("Forme avant preprocessing:", deep_features.head(20).shape)
print("Forme aprÃ¨s preprocessing:", transformed_preview.shape)
print("Noms gÃ©nÃ©rÃ©s:", deep_preprocessor.get_feature_names_out())
````

## 3. Construction du MLP et callbacks

- `EarlyStopping` arrÃªte l'entraÃ®nement si `val_loss` ne s'amÃ©liore plus et restaure les meilleurs poids.
- `ReduceLROnPlateau` rÃ©duit le learning rate lorsque l'optimisation stagne.
- Le constructeur reÃ§oit automatiquement le nombre de features aprÃ¨s preprocessing.

````python
def build_demand_mlp(input_size: int, output_size: int) -> tf.keras.Model:
    """Construit le rÃ©seau utilisÃ© par cet exemple de prÃ©vision de demande."""
    model = tf.keras.Sequential(
        [
            tf.keras.layers.Input(shape=(input_size,)),
            tf.keras.layers.Dense(64, activation="relu"),
            tf.keras.layers.Dropout(0.15),
            tf.keras.layers.Dense(32, activation="relu"),
            tf.keras.layers.Dense(1),
        ],
        name="avenqo_demand_mlp",
    )
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss="mean_squared_error",
        metrics=["mean_absolute_error"],
    )
    return model


early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=8,
    restore_best_weights=True,
)
reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=0.00001,
)

print("Callbacks prÃªts :", early_stopping.__class__.__name__, reduce_lr.__class__.__name__)
````

````python
deep_result = training_service.train_neural_regressor(
    data=demand_data,
    target_column="demand",
    dataset=DATASET,
    version="demand-mlp-v1",
    run_context=build_run_context(demand_data),
    model_builder=build_demand_mlp,
    callbacks=[early_stopping, reduce_lr],
    hidden_units=(64, 32),
    learning_rate=0.001,
    epochs=60,
    batch_size=32,
    test_size=0.2,
    random_seed=RANDOM_SEED,
)

print("Ã‰poques rÃ©ellement exÃ©cutÃ©es:", len(deep_result.history["loss"]))
print("MÃ©triques de test:", dict(deep_result.metrics))
deep_result.model.summary()
````

````python
history = pd.DataFrame(deep_result.history)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
history[["loss", "val_loss"]].plot(ax=axes[0], color=["#167d8d", "#d97941"])
axes[0].set_title("Erreur quadratique pendant l'entraÃ®nement")
axes[0].set_xlabel("Ã‰poque")
axes[0].set_ylabel("MSE")

history[["mean_absolute_error", "val_mean_absolute_error"]].plot(
    ax=axes[1], color=["#2f6b4f", "#c84b31"]
)
axes[1].set_title("Erreur absolue moyenne")
axes[1].set_xlabel("Ã‰poque")
axes[1].set_ylabel("MAE")
plt.tight_layout()
plt.show()
````

## 4. Sauvegarde, rechargement et prÃ©diction

Le fichier `.keras` contient le rÃ©seau. Le fichier `preprocessor.joblib` contient les transformations numÃ©riques et catÃ©gorielles. En production, les deux doivent toujours Ãªtre chargÃ©s ensemble.

````python
loaded_model = tf.keras.models.load_model(deep_result.model_path)
loaded_preprocessor = joblib.load(deep_result.preprocessor_path)

new_order = pd.DataFrame(
    [{"price": 25.0, "promotion": 0.20, "channel": "web", "season": "summer"}]
)
prepared_order = loaded_preprocessor.transform(new_order)
predicted_demand = float(loaded_model.predict(prepared_order, verbose=0)[0, 0])

print("ModÃ¨le sauvegardÃ©       :", deep_result.model_path)
print("PrÃ©processeur sauvegardÃ©:", deep_result.preprocessor_path)
print(f"Demande prÃ©dite          : {predicted_demand:.2f}")
````
